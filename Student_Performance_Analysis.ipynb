```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("Libraries imported successfully!")
```

```python
data = {
    'Student_ID': list(range(101, 131)),
    'Name': [
        'Rahul', 'Priya', 'Aman', 'Sneha', 'Karan',
        'Anjali', 'Rohit', 'Neha', 'Arjun', 'Simran',
        'Vikas', 'Pooja', 'Amit', 'Kirti', 'Varun',
        'Riya', 'Manish', 'Shivani', 'Deepak', 'Meena',
        'Ankit', 'Komal', 'Sahil', 'Divya', 'Mohit',
        'Tanya', 'Ajay', 'Nisha', 'Sumit', 'Muskan'
    ],
    'Study_Hours': [
        3, 5, 2, 6, 4, 5, 2, 7, 3, 6,
        4, 5, 1, 7, 3, 6, 2, 5, 4, 7,
        3, 6, 2, 5, 4, 7, 1, 6, 3, 5
    ],
    'Attendance': [
        80, 90, 65, 95, 75, 85, 60, 98, 70, 90,
        80, 85, 55, 95, 75, 92, 65, 88, 78, 96,
        72, 91, 68, 89, 82, 97, 50, 93, 74, 87
    ],
    'Marks': [
        65, 85, 50, 92, 72, 80, 45, 95, 60, 88,
        75, 82, 35, 94, 62, 90, 48, 84, 70, 93,
        58, 87, 52, 86, 76, 96, 30, 89, 63, 83
    ]
}

df = pd.DataFrame(data)

df.to_csv('students.csv', index=False)

df.head()
```

```python
# Check dataset information
df.info()
```

```python
# Statistical summary
df.describe()
```

```python
# Average marks
average_marks = df['Marks'].mean()

# Highest marks
highest_marks = df['Marks'].max()

# Lowest marks
lowest_marks = df['Marks'].min()

# Average study hours
average_study = df['Study_Hours'].mean()

# Average attendance
average_attendance = df['Attendance'].mean()

print("STUDENT PERFORMANCE ANALYSIS")
print("-----------------------------")
print("Average Marks:", round(average_marks, 2))
print("Highest Marks:", highest_marks)
print("Lowest Marks:", lowest_marks)
print("Average Study Hours:", round(average_study, 2))
print("Average Attendance:", round(average_attendance, 2), "%")
```

```python
# Find top 5 students
top_students = df.sort_values(by='Marks', ascending=False).head(5)

print("TOP 5 PERFORMING STUDENTS")
top_students[['Name', 'Study_Hours', 'Attendance', 'Marks']]
```

```python
plt.figure(figsize=(12, 5))

plt.bar(df['Name'], df['Marks'])

plt.title('Student Performance Analysis')
plt.xlabel('Student Name')
plt.ylabel('Marks')

plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(df['Study_Hours'], df['Marks'])

plt.title('Study Hours vs Student Marks')
plt.xlabel('Study Hours')
plt.ylabel('Marks')

plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(df['Attendance'], df['Marks'])

plt.title('Attendance vs Student Marks')
plt.xlabel('Attendance (%)')
plt.ylabel('Marks')

plt.grid(True)
plt.show()
```

```python
# Calculate correlation
correlation = df[['Study_Hours', 'Attendance', 'Marks']].corr()

print("CORRELATION ANALYSIS")
print(correlation.round(2))
```
